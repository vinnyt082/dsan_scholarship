# 00 Project Scope

Working title: **Trading Time for Affordability**

This notebook records the project scope before ingestion, analysis, visualization, or app development begins.

## Core Thesis

Housing affordability should not be understood only as the sticker price of housing. A home's real affordability also depends on access: how much time, distance, and transportation burden is required to reach work and participate in daily economic life.

The final project should tell a narrow, polished, data-driven story about the combined burden of housing and access in the United States.

## Research Question

**Has the pursuit of affordable housing increasingly required Americans to pay in commute time and access?**

## Claim Style

Use careful, non-causal language. The project may describe patterns as associated with, coinciding with, suggesting, being consistent with, or revealing a tradeoff.

Do not claim that housing prices directly caused commute times to rise. Do not claim that the analysis proves suburbanization.

## Final Product

The target product is a Streamlit data story, not a generic dashboard. The final app should feel like a guided visual article with polished charts, concise text, and strong visual hierarchy.

Expected structure:

1. Home / Story Overview
2. Housing Prices
3. Paychecks and Purchasing Power
4. Commute Burden
5. Housing + Access
6. Methods / References

## Non-Goals

- Do not build a broad cost-of-living dashboard.
- Do not add excessive variables.
- Do not scrape websites.
- Do not make causal claims.
- Do not attempt every U.S. county or metro unless specifically instructed.
- Do not build Streamlit pages before processed data and visual prototypes are reviewed.

## Planned Data

Primary source should be FRED where possible. Candidate FRED series include housing prices, CPI/inflation, home price index, mortgage rates, real median household income, shelter/rent CPI, and transportation services CPI.

Commute/access data should start with an inventory of FRED-hosted ACS mean commute-time series where possible. If FRED is not sufficient, later work may evaluate Census ACS data in a controlled way. Census API ingestion should not be implemented until explicitly instructed.

## Geographic Scope

Use national data for long-run housing, inflation, income, and mortgage context. For commute/access, start with national commute trend if available. Selected metros or counties may be considered only after data availability is checked.

## Analytical Priorities

The project should measure:

1. Housing sticker-price pressure.
2. Income or purchasing-power context.
3. Commute/access burden.
4. The combined burden of housing plus access.

A likely final visual centerpiece is a housing + access quadrant chart with housing affordability pressure on the x-axis and mean commute time on the y-axis.

## Workflow

The analytical workflow is notebook-first for direct control and editing. The Streamlit app should read processed CSVs created by notebooks. Helper scripts are allowed only for reusable functions.

Current phase: scaffold and planning only.
